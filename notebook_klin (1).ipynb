```python
# TS1.2.1. Mostrar las transacciones registradas en la tabla transacciones_servicios, calculando para cada una el importe con un IVA del 21%.

query = """
SELECT
    transaccion_id,
    cliente_id,
    servicio_id,
    fecha_hora,
    importe,
    ROUND(importe * 1.21, 2) AS importe_con_iva
FROM transacciones_servicios
ORDER BY fecha_hora
LIMIT 20;
"""

cursor.execute(query)
resultados = cursor.fetchall()

for fila in resultados:
    print(fila)
```

```python
# TS1.2.2. Número de transacciones por estado. Obtener cuántas transacciones hay en cada estado (completada, pendiente, cancelada, fallida).

query = """
SELECT
    estado,
    COUNT(*) AS total_transacciones
FROM transacciones_servicios
GROUP BY estado
ORDER BY total_transacciones DESC;
"""

cursor.execute(query)
resultados = cursor.fetchall()

for fila in resultados:
    print(fila)
```

```python
# TS1.2.3. Gasto total de cada cliente. Obtener el gasto total realizado por cada cliente en transacciones de servicios.

query = """
SELECT
    c.cliente_id,
    c.nombre,
    c.apellidos,
    COUNT(ts.transaccion_id) AS total_transacciones,
    SUM(ts.importe) AS gasto_total
FROM clientes c
JOIN transacciones_servicios ts
    ON c.cliente_id = ts.cliente_id
GROUP BY c.cliente_id, c.nombre, c.apellidos
ORDER BY gasto_total DESC
LIMIT 25;
"""

cursor.execute(query)
resultados = cursor.fetchall()

for fila in resultados:
    print(fila)
```

```python
# TS1.2.4. Número de clientes captados por cada canal. Obtener cuántos clientes se han captado por cada canal de captación.

query = """
SELECT
    cc.canal_id,
    cc.nombre_canal,
    cc.tipo_canal,
    COUNT(c.cliente_id) AS total_clientes
FROM canales_captacion cc
LEFT JOIN clientes c
    ON cc.canal_id = c.canal_id
GROUP BY cc.canal_id, cc.nombre_canal, cc.tipo_canal
ORDER BY total_clientes DESC;
"""

cursor.execute(query)
resultados = cursor.fetchall()

for fila in resultados:
    print(fila)
```

```python
# TS1.3.1

query = """
SELECT
    transaccion_id,
    importe
FROM transacciones_servicios
ORDER BY fecha_hora
LIMIT 20;
"""

cursor.execute(query)
resultados = cursor.fetchall()

contador_bajo = 0
contador_medio = 0
contador_alto = 0

for fila in resultados:
    transaccion_id = fila[0]
    importe = fila[1]

    if importe < 20:
        categoria = "Bajo"
        contador_bajo += 1
    elif importe <= 50:
        categoria = "Medio"
        contador_medio += 1
    else:
        categoria = "Alto"
        contador_alto += 1

    print(f"Transacción {transaccion_id}: importe = {importe} € -> categoría {categoria}")

print("\nResumen final:")
print(f"Transacciones bajas: {contador_bajo}")
print(f"Transacciones medias: {contador_medio}")
print(f"Transacciones altas: {contador_alto}")
```

```python
# TS1.3.2

query = """
SELECT
    transaccion_id,
    estado,
    importe
FROM transacciones_servicios
ORDER BY fecha_hora
LIMIT 30;
"""

cursor.execute(query)
resultados = cursor.fetchall()

contador_ok = 0
contador_problematicas = 0

for fila in resultados:
    transaccion_id = fila[0]
    estado = fila[1]
    importe = fila[2]

    # lógica de detección
    if (
        estado == "fallida" or
        estado == "cancelada" or
        (estado == "pendiente" and importe > 50)
    ):
        clasificacion = "PROBLEMÁTICA"
        contador_problematicas += 1
    else:
        clasificacion = "OK"
        contador_ok += 1

    print(f"Transacción {transaccion_id} | Estado: {estado} | Importe: {importe} € -> {clasificacion}")

print("\nResumen final:")
print(f"Transacciones OK: {contador_ok}")
print(f"Transacciones problemáticas: {contador_problematicas}")
```

```python
# TS1.3.3

import time

# 1. Obtener servicios activos
query = """
SELECT nombre_servicio
FROM servicios
WHERE activo = TRUE
ORDER BY servicio_id
LIMIT 5;
"""

cursor.execute(query)
servicios_db = cursor.fetchall()

servicios = [fila[0] for fila in servicios_db]

# 2. Tiempo total en segundos
tiempo_restante = 300  # 5 minutos

print("🚗 Iniciando lavado KLIN...\n")

# 3. Simulación
while tiempo_restante > 0:

    # Cada 60 segundos cambia el servicio
    minuto_actual = (300 - tiempo_restante) // 60
    servicio_actual = servicios[minuto_actual % len(servicios)]

    print(f"⏳ Tiempo restante: {tiempo_restante} segundos")
    print(f"🔧 Servicio en ejecución: {servicio_actual}")
    print("-" * 40)

    time.sleep(1)
    tiempo_restante -= 1

print("\n✅ Lavado finalizado. ¡Gracias por usar KLIN!")
```

```python
import time


# --------------------------------------------------
# 1. FUNCIÓN DE NEGOCIO:
#    clasifica un importe en Bajo, Medio o Alto
# --------------------------------------------------
def clasificar_importe(importe):
    """
    Clasifica un importe según estas reglas:
    - Bajo: importe < 20
    - Medio: 20 <= importe <= 50
    - Alto: importe > 50

    Parámetros:
        importe (float o int): cantidad a clasificar

    Devuelve:
        str: categoría del importe
    """

    # Validación básica para evitar valores no válidos
    if importe < 0:
        raise ValueError("El importe no puede ser negativo")

    if importe < 20:
        return "Bajo"
    elif importe <= 50:
        return "Medio"
    else:
        return "Alto"


# --------------------------------------------------
# 2. FUNCIÓN DE SIMULACIÓN:
#    simula un lavado corto mostrando servicios
# --------------------------------------------------
def ejecutar_simulacion_lavado(servicios, duracion_total=15, intervalo_cambio=5):
    """
    Simula una pantalla de lavado.

    Parámetros:
        servicios (list): lista de nombres de servicios
        duracion_total (int): duración total en segundos
        intervalo_cambio (int): cada cuántos segundos cambia el servicio

    Devuelve:
        list: historial de mensajes generados durante la simulación
    """

    # Si no hay servicios, la simulación no puede ejecutarse
    if not servicios:
        raise ValueError("La lista de servicios no puede estar vacía")

    historial = []

    print("🚗 Iniciando simulación de lavado...\n")

    for segundo_restante in range(duracion_total, 0, -1):
        # Calculamos qué tramo de servicio corresponde
        tramo_actual = (duracion_total - segundo_restante) // intervalo_cambio
        servicio_actual = servicios[tramo_actual % len(servicios)]

        mensaje = (
            f"Tiempo restante: {segundo_restante:02d} s | "
            f"Servicio en ejecución: {servicio_actual}"
        )

        print(mensaje)
        historial.append(mensaje)

        # Espera de 1 segundo para simular el paso del tiempo
        time.sleep(1)

    print("\n✅ Lavado finalizado")
    return historial


# --------------------------------------------------
# 3. PRUEBAS AUTOMÁTICAS BÁSICAS
#    comprobamos que la lógica funciona
# --------------------------------------------------
def ejecutar_tests():
    """
    Ejecuta una batería sencilla de pruebas automáticas.
    Si alguna falla, Python lanzará un AssertionError.
    """

    print("🧪 Ejecutando tests...")

    # ---- Tests de clasificación de importe ----
    # Casos normales
    assert clasificar_importe(10) == "Bajo", "10 debería clasificarse como Bajo"
    assert clasificar_importe(30) == "Medio", "30 debería clasificarse como Medio"
    assert clasificar_importe(80) == "Alto", "80 debería clasificarse como Alto"

    # Casos frontera (muy importantes en testing)
    assert clasificar_importe(19.99) == "Bajo", "19.99 debería ser Bajo"
    assert clasificar_importe(20) == "Medio", "20 debería ser Medio"
    assert clasificar_importe(50) == "Medio", "50 debería ser Medio"
    assert clasificar_importe(50.01) == "Alto", "50.01 debería ser Alto"

    # Test de error esperado
    try:
        clasificar_importe(-5)
        assert False, "Se esperaba un ValueError para importes negativos"
    except ValueError:
        pass

    # ---- Tests de simulación ----
    servicios_prueba = ["Prelavado", "Espuma", "Aclarado"]

    historial = ejecutar_simulacion_lavado(
        servicios=servicios_prueba,
        duracion_total=3,
        intervalo_cambio=1
    )

    # Comprobamos que se han generado 3 líneas de simulación
    assert len(historial) == 3, "La simulación debería generar 3 mensajes"

    # Comprobamos que aparecen los servicios esperados
    assert "Prelavado" in historial[0], "El primer servicio debería ser Prelavado"
    assert "Espuma" in historial[1], "El segundo servicio debería ser Espuma"
    assert "Aclarado" in historial[2], "El tercer servicio debería ser Aclarado"

    print("✅ Todos los tests han pasado correctamente\n")


# --------------------------------------------------
# 4. FUNCIÓN PRINCIPAL:
#    orquesta todo el flujo completo
# --------------------------------------------------
def main():
    """
    Ejecuta el flujo completo:
    1. lanza tests
    2. ejecuta una simulación
    3. clasifica un importe
    4. muestra un resumen final
    """

    # Primero validamos que la lógica es correcta
    ejecutar_tests()

    # Lista de servicios que verá el usuario durante el lavado
    servicios_reales = [
        "Prelavado",
        "Espuma activa",
        "Cepillos",
        "Aclarado",
        "Secado"
    ]

    # Ejecutamos una simulación de ejemplo
    ejecutar_simulacion_lavado(
        servicios=servicios_reales,
        duracion_total=10,
        intervalo_cambio=2
    )

    # Clasificamos un importe de ejemplo
    importe = 47.50
    categoria = clasificar_importe(importe)

    # Mostramos resultado final
    print("\n📋 Resumen final del proceso")
    print(f"Importe analizado: {importe} €")
    print(f"Categoría asignada: {categoria}")
    print("Proceso completado correctamente.")


# --------------------------------------------------
# 5. PUNTO DE ENTRADA DEL PROGRAMA
# --------------------------------------------------
if __name__ == "__main__":
    main()
```
