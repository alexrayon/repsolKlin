```python
# TS1.2.2. Número de transacciones por estado. Obtener cuántas transacciones hay en cada estado (completada, pendiente, cancelada, fallida).

query = """
SELECT
    estado,
    COUNT(*) AS total_transacciones
FROM transacciones_servicios
GROUP BY estado
ORDER BY total_transacciones DESC;
"""

cursor.execute(query)
resultados = cursor.fetchall()

for fila in resultados:
    print(fila)
```

```python
# TS1.2.3. Gasto total de cada cliente. Obtener el gasto total realizado por cada cliente en transacciones de servicios.

query = """
SELECT
    c.cliente_id,
    c.nombre,
    c.apellidos,
    COUNT(ts.transaccion_id) AS total_transacciones,
    SUM(ts.importe) AS gasto_total
FROM clientes c
JOIN transacciones_servicios ts
    ON c.cliente_id = ts.cliente_id
GROUP BY c.cliente_id, c.nombre, c.apellidos
ORDER BY gasto_total DESC
LIMIT 25;
"""

cursor.execute(query)
resultados = cursor.fetchall()

for fila in resultados:
    print(fila)
```

```python
# TS1.2.4. Número de clientes captados por cada canal. Obtener cuántos clientes se han captado por cada canal de captación.

query = """
SELECT
    cc.canal_id,
    cc.nombre_canal,
    cc.tipo_canal,
    COUNT(c.cliente_id) AS total_clientes
FROM canales_captacion cc
LEFT JOIN clientes c
    ON cc.canal_id = c.canal_id
GROUP BY cc.canal_id, cc.nombre_canal, cc.tipo_canal
ORDER BY total_clientes DESC;
"""

cursor.execute(query)
resultados = cursor.fetchall()

for fila in resultados:
    print(fila)
```

```python
# TS1.3.1

query = """
SELECT
    transaccion_id,
    importe
FROM transacciones_servicios
ORDER BY fecha_hora
LIMIT 20;
"""

cursor.execute(query)
resultados = cursor.fetchall()

contador_bajo = 0
contador_medio = 0
contador_alto = 0

for fila in resultados:
    transaccion_id = fila[0]
    importe = fila[1]

    if importe < 20:
        categoria = "Bajo"
        contador_bajo += 1
    elif importe <= 50:
        categoria = "Medio"
        contador_medio += 1
    else:
        categoria = "Alto"
        contador_alto += 1

    print(f"Transacción {transaccion_id}: importe = {importe} € -> categoría {categoria}")

print("\nResumen final:")
print(f"Transacciones bajas: {contador_bajo}")
print(f"Transacciones medias: {contador_medio}")
print(f"Transacciones altas: {contador_alto}")
```

```python
# TS1.3.2

query = """
SELECT
    transaccion_id,
    estado,
    importe
FROM transacciones_servicios
ORDER BY fecha_hora
LIMIT 30;
"""

cursor.execute(query)
resultados = cursor.fetchall()

contador_ok = 0
contador_problematicas = 0

for fila in resultados:
    transaccion_id = fila[0]
    estado = fila[1]
    importe = fila[2]

    # lógica de detección
    if (
        estado == "fallida" or
        estado == "cancelada" or
        (estado == "pendiente" and importe > 50)
    ):
        clasificacion = "PROBLEMÁTICA"
        contador_problematicas += 1
    else:
        clasificacion = "OK"
        contador_ok += 1

    print(f"Transacción {transaccion_id} | Estado: {estado} | Importe: {importe} € -> {clasificacion}")

print("\nResumen final:")
print(f"Transacciones OK: {contador_ok}")
print(f"Transacciones problemáticas: {contador_problematicas}")
```

```python
# TS1.3.3

import time

# 1. Obtener servicios activos
query = """
SELECT nombre_servicio
FROM servicios
WHERE activo = TRUE
ORDER BY servicio_id
LIMIT 5;
"""

cursor.execute(query)
servicios_db = cursor.fetchall()

servicios = [fila[0] for fila in servicios_db]

# 2. Tiempo total en segundos
tiempo_restante = 300  # 5 minutos

print("🚗 Iniciando lavado KLIN...\n")

# 3. Simulación
while tiempo_restante > 0:

    # Cada 60 segundos cambia el servicio
    minuto_actual = (300 - tiempo_restante) // 60
    servicio_actual = servicios[minuto_actual % len(servicios)]

    print(f"⏳ Tiempo restante: {tiempo_restante} segundos")
    print(f"🔧 Servicio en ejecución: {servicio_actual}")
    print("-" * 40)

    time.sleep(1)
    tiempo_restante -= 1

print("\n✅ Lavado finalizado. ¡Gracias por usar KLIN!")
```

```python
# 🔹 PASO. Detectar celdas que pueden dar problemas antes de subir a GitHub

from google.colab import _message
import re

# Obtengo el notebook que estoy ejecutando actualmente
notebook_actual = _message.blocking_request(
    "get_ipynb",
    timeout_sec=600
)["ipynb"]

# Patrones que queremos detectar
patrones = {
    "GitHub token": r"github_pat_[A-Za-z0-9_]+",
    "GitHub token clásico": r"ghp_[A-Za-z0-9]+",
    "Password": r"password\s*=",
    "Secret": r"secret\s*=",
    "API key": r"api[_-]?key\s*="
}

problemas = []

# Recorremos todas las celdas
for numero, celda in enumerate(notebook_actual["cells"], start=1):

    contenido = celda.get("source", "")

    if isinstance(contenido, list):
        contenido = "".join(contenido)

    for nombre, patron in patrones.items():

        if re.search(patron, contenido, re.IGNORECASE):

            problemas.append(
                (numero, nombre, contenido[:150])
            )


# Mostramos el resultado
if not problemas:

    print("✓ No se han encontrado celdas sospechosas")

else:

    print("⚠️ Se han encontrado posibles problemas:\n")

    for numero, tipo, contenido in problemas:

        print(f"Celda {numero}")
        print(f"Problema detectado: {tipo}")
        print("Contenido:")
        print(contenido)
        print("-" * 50)
```
